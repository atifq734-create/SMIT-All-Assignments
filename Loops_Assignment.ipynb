## Student Marks Entry

### Goal:
    Record student marks and skip invalid entries.

### Steps:

    Create an empty list marks = [].
    
    Ask user to enter marks for 5 students.
    
    If marks < 0 or > 100 → print “Invalid marks” and continue.
    
    If user enters -1 → stop taking input (break).
    
    Print valid marks and their average.

### Example:

    Enter marks: 90  
    Enter marks: 105 → Invalid marks  
    Enter marks: 80  
    Enter marks: -1 → stop  
    Valid Marks: [90, 80]  
    Average = 85.0

```python
marks = []

for i in range(5):
    mark = int(input("Enter student marks (-1 to stop): "))

    if mark == -1:
        break

    if mark < 0 or mark > 100:
        print("Invalid marks")
        continue

    marks.append(mark)

print("\nValid marks:", marks)

if len(marks) > 0:
    average = sum(marks) / len(marks)
    print("Average marks:", average)
else:
    print("No valid marks entered.")
```

## 🏨 8. Hotel Room Booking

### Goal:
    Manage simple room booking.

### Task:

    Create a list of rooms [101, 102, 103, 104, 105]
    
    Ask user to choose a room.
    
    If room already booked, print “Room not available.”
    
    Continue until all rooms are booked.

### Example:

    Available rooms: [101, 102, 103, 104, 105]
    Choose room: 103
    Room 103 booked successfully!

```python
rooms = [101, 102, 103, 104, 105]
booked_rooms = []

while len(booked_rooms) < len(rooms):
    print("Available rooms:", rooms)

    room = int(input("Choose a room to book: "))

    if room not in rooms:
        print("Invalid room number.")
        continue

    if room in booked_rooms:
        print("Room not available.")
        continue

    booked_rooms.append(room)
    rooms.remove(room)

    print("Room", room, "booked successfully!")

print("\nAll rooms are booked.")
print("Booked rooms:", booked_rooms)
```

### Goal:
    Record 5 transactions and skip invalid amounts.

### Steps:

    Ask the user to enter deposit amounts (positive integers).

## Car Rental System

### Goal:
    Create a car rental booking system.

### Steps:

    Cars = ["Civic", "Corolla", "City", "Alto"]
    
    Use a for loop to let multiple users book.
    
    If user enters "stop" → break.
    
    If car not in list → continue (invalid).
    
    Add valid bookings to booked_cars list.

### Example:

    Enter car name: Civic  
    Enter car name: Alto  
    Enter car name: Tesla → Not available  
    Enter car name: stop  
    Booked Cars: ['Civic', 'Alto']

```python
cars = ["Civic", "Corolla", "City", "Alto"]
booked_cars = []

for i in range(100):
    print("\nAvailable cars:", cars)
    car = input("Enter car name to book (or 'stop' to finish): ")

    if car.lower() == "stop":
        break

    found = False

    for available_car in cars:
        if car.lower() == available_car.lower():
            found = True
            booked_cars.append(available_car)
            print(available_car, "booked successfully!")
            break

    if not found:
        print("Invalid car. Please try again.")
        continue

print("\nBooked Cars:")
for car in booked_cars:
    print(car)

print("Total bookings:", len(booked_cars))
```

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python

balance = 0
account_created = False

while True:
    print("\n===== Banking System =====")
    print("1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check Balance")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        if account_created:
            print("Account already exists.")
        else:
            name = input("Enter your name: ")
            account_created = True
            print("Account created successfully!")
            print("Account holder:", name)

    elif choice == "2":
        if not account_created:
            print("Please create an account first.")
            continue

        amount = float(input("Enter deposit amount: "))

        if amount <= 0:
            print("Invalid deposit amount.")
        else:
            balance += amount
            print("Amount deposited successfully!")

    elif choice == "3":
        if not account_created:
            print("Please create an account first.")
            continue

        amount = float(input("Enter withdrawal amount: "))

        if amount <= 0:
            print("Invalid withdrawal amount.")
        elif amount > balance:
            print("Insufficient balance.")
        else:
            balance -= amount
            print("Withdrawal successful!")

    elif choice == "4":
        if not account_created:
            print("Please create an account first.")
            continue

        print("Your current balance is:", balance)

    elif choice == "5":
        print("Thank you for using our banking system!")
        break

    else:
        print("Invalid choice. Please try again.")
```

## School Marks Report
### Goal:

    Take input for multiple students’ marks in 3 subjects and calculate their total and average.

### Steps:

    Outer loop → for each student (3 students).
    
    Inner loop → take marks for 3 subjects.
    
    Store each student’s total in a list.
    
    Print all totals and class average.

### Example:
    Enter marks for Student 1:
    Subject 1: 80
    Subject 2: 75
    Subject 3: 90
    Enter marks for Student 2:
    Subject 1: 60
    Subject 2: 70
    Subject 3: 65
    → Output: [245, 195]

```python

totals = []

for i in range(3):
    total = 0
    print(f"\nStudent {i + 1}")

    for j in range(3):
        marks = float(input(f"Enter marks for subject {j + 1}: "))
        total += marks

    totals.append(total)
    average = total / 3

    print("Total marks:", total)
    print("Average marks:", average)

print("\n===== School Marks Report =====")
for i in range(3):
    print(f"Student {i + 1} Total: {totals[i]}")

class_average = sum(totals) / len(totals)
print("Class Average:", class_average)
```

## Student Attendance Tracker
### Goal:

    Record attendance (Present/Absent) for multiple classes and students.

### Steps:

    Outer loop → each class (2 classes).
    
    Inner loop → each student (3 students).
    
    Store "P" or "A" in a nested list.

    Print full attendance chart.

### Example:
    Class 1:
    Student 1 (P/A): P
    Student 2 (P/A): A
    Student 3 (P/A): P
    Class 2:
    Student 1 (P/A): P
    Student 2 (P/A): P
    Student 3 (P/A): P
    Output: [['P', 'A', 'P'], ['P', 'P', 'P']]


```python

attendance = []

for class_num in range(1, 3):
    print(f"\nClass {class_num}")
    class_attendance = []

    for student in range(1, 4):
        status = input(
            f"Enter attendance for Student {student} (P/A): "
        ).upper()

        while status not in ["P", "A"]:
            print("Invalid input! Enter P or A.")
            status = input(
                f"Enter attendance for Student {student} (P/A): "
            ).upper()

        class_attendance.append(status)

    attendance.append(class_attendance)

print("\n===== Attendance Chart =====")

for i in range(2):
    print(f"Class {i + 1}:", attendance[i])
```

## Cinema Seat Booking System
### Goal:

    Take user input for seat availability across rows and columns, and mark booked seats.

### Steps:

    Outer loop → for each row (3 rows).
    
    Inner loop → for 5 seats in that row.
    
    Ask if seat is booked (Y/N).
    
    If booked → add “Booked” else “Available”.
    
    Store in nested list and print seating chart.

### Example:
    Row 1:
    Seat 1 (Y/N): Y
    Seat 2 (Y/N): N
    Seat 3 (Y/N): Y
    Seat 4 (Y/N): N
    Seat 5 (Y/N): N
    Row 2:
    Seat 1 (Y/N): Y
    Seat 2 (Y/N): Y
    Seat 3 (Y/N): Y
    Seat 4 (Y/N): Y
    Seat 5 (Y/N): N
    
    Output:
    [['Booked','Available','Booked','Available','Available'],
     ['Booked','Booked','Booked','Booked','Available']]

```python

seating_chart = []

for row in range(1, 4):
    print(f"\nRow {row}")
    seats = []

    for seat in range(1, 6):
        status = input(
            f"Is seat {seat} booked? (Y/N): "
        ).upper()

        if status == "Y":
            seats.append("Booked")
        elif status == "N":
            seats.append("Available")
        else:
            print("Invalid input! Seat marked as Available.")
            seats.append("Available")

    seating_chart.append(seats)

print("\n===== Cinema Seating Chart =====")

for i in range(3):
    print(f"Row {i + 1}:", seating_chart[i])
```

## Hotel Room Booking Tracker
### Goal:

    Record bookings for rooms on each floor.

### Steps:

    Outer loop → for 3 floors.
    
    Inner loop → for 4 rooms per floor.
    
    Ask user if booked (Y/N).
    
    Count total booked and available rooms.
    
    Print summary per floor.

### Example:
    Floor 1:
    Room 1: Y
    Room 2: N
    Room 3: N
    Room 4: Y
    Floor 2:
    Room 1: Y
    Room 2: Y
    Room 3: Y
    Room 4: Y
    → Output: Floor 1 = 2 booked, Floor 2 = 4 booked

```python

for floor in range(1, 4):
    booked = 0
    available = 0

    print(f"\nFloor {floor}")

    for room in range(1, 5):
        status = input(
            f"Is room {room} booked? (Y/N): "
        ).upper()

        if status == "Y":
            booked += 1
        elif status == "N":
            available += 1
        else:
            print("Invalid input! Room counted as available.")
            available += 1

    print("Booked rooms:", booked)
    print("Available rooms:", available)

print("\nHotel booking tracking completed.")
```

```python

for invoice in range(1, 4):
    items = []
    total = 0

    print(f"\n===== Invoice {invoice} =====")

    for item in range(1, 5):
        price = float(input(f"Enter price of item {item}: "))
        items.append(price)
        total += price

    tax = total * 0.10
    final_total = total + tax

    print("\nItems prices:", items)
    print("Subtotal:", total)
    print("Tax (10%):", tax)
    print("Invoice Total:", final_total)
```

## Student Result Sheet
### Goal:

    Generate report cards for each student in a class.

### Steps:

    Outer loop → 5 students.
    
    Inner loop → 4 subjects.

    If marks < 33 → print “Failed in Subject”.

    Calculate total and grade per student.

### Example:
    Student 1: [80, 90, 70, 85] → Grade: A  
    Student 2: [50, 32, 45, 40] → Failed in Subject 2

```python

for student in range(1, 6):
    total = 0
    failed = False

    print(f"\n===== Student {student} Result =====")

    for subject in range(1, 5):
        marks = int(input(f"Enter marks for Subject {subject}: "))

        total += marks

        if marks < 33:
            print(f"Failed in Subject {subject}")
            failed = True

    average = total / 4

    if failed:
        grade = "F"
    elif average >= 80:
        grade = "A"
    elif average >= 70:
        grade = "B"
    elif average >= 60:
        grade = "C"
    elif average >= 50:
        grade = "D"
    else:
        grade = "F"

    print("Total Marks:", total, "/ 400")
    print("Average:", average)
    print("Grade:", grade)

    if failed:
        print("Result: FAIL")
    else:
        print("Result: PASS")
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python

features = ["Hotel", "Transport", "Food"]

for package in range(1, 4):
    total_rating = 0
    count = 0

    print(f"\n===== Package {package} =====")

    for feature in features:
        rating = int(input(f"Rate {feature} (0-5): "))

        if rating == 0:
            continue

        if rating < 0 or rating > 5:
            print("Invalid rating! Enter a value from 0 to 5.")
            continue

        total_rating += rating
        count += 1

    if count > 0:
        average = total_rating / count
        print("Average Rating:", round(average, 2), "/ 5")
    else:
        print("No ratings provided.")
```
