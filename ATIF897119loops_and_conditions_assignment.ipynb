## ⚖️ 2. BMI Health Checker

### Goal:
    Help a fitness app tell users which weight category they belong to.

### Steps:

    Ask for weight in kg and height in metres.
    
    Calculate BMI = weight / (height * height).
    
    BMI below 18.5 → "Underweight".
    
    18.5 to 24.9 → "Normal".
    
    25 to 29.9 → "Overweight".
    
    30 or more → "Obese".
    
    Print the BMI rounded to 2 decimals.

### Example:

    Enter weight: 70
    Enter height: 1.75
    BMI = 22.86 → Normal

```python
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in metres: "))

bmi = weight / (height * height)

print("BMI:", round(bmi, 2))

if bmi < 18.5:
    print("Underweight")
elif bmi < 25:
    print("Normal")
elif bmi < 30:
    print("Overweight")
else:
    print("Obese")
```

## 🌡️ 8. Temperature Unit Converter

### Goal:
    Build a small converter for a weather app.

### Steps:

    Show a menu: 1 = Celsius to Fahrenheit, 2 = Fahrenheit to Celsius.
    
    Ask for the choice and then the temperature.
    
    Choice 1 → F = C * 9/5 + 32.
    
    Choice 2 → C = (F - 32) * 5/9.
    
    Any other choice → "Invalid choice".

### Example:

    Enter choice: 1
    Enter temperature: 100
    212.0 F

```python
print("Temperature Converter")
print("1 = Celsius to Fahrenheit")
print("2 = Fahrenheit to Celsius")

choice = int(input("Enter your choice (1 or 2): "))

if choice == 1:
    celsius = float(input("Enter temperature in Celsius: "))
    fahrenheit = celsius * 9/5 + 32
    print("Temperature in Fahrenheit:", fahrenheit)

elif choice == 2:
    fahrenheit = float(input("Enter temperature in Fahrenheit: "))
    celsius = (fahrenheit - 32) * 5/9
    print("Temperature in Celsius:", celsius)

else:
    print("Invalid choice")
```

## ✊ 17. Rock, Paper, Scissors Game

### Goal:
    Decide the winner of a two-player game.

### Steps:

    Ask Player 1 and Player 2 for rock, paper or scissors.
    
    Rock beats scissors, scissors beats paper, paper beats rock.
    
    Use logical operators (and / or) to print the winner.
    
    If both choose the same → "Draw".

### Example:

    Player 1: rock
    Player 2: scissors
    Player 1 wins

```python
player1 = input("Player 1, choose rock, paper, or scissors: ").lower()
player2 = input("Player 2, choose rock, paper, or scissors: ").lower()

if player1 == player2:
    print("Draw")

elif (player1 == "rock" and player2 == "scissors") or \
     (player1 == "scissors" and player2 == "paper") or \
     (player1 == "paper" and player2 == "rock"):
    print("Player 1 wins")

elif (player2 == "rock" and player1 == "scissors") or \
     (player2 == "scissors" and player1 == "paper") or \
     (player2 == "paper" and player1 == "rock"):
    print("Player 2 wins")

else:
    print("Invalid choice")
```

## 🌡️ 50. Weekly Temperature Report

### Goal:
    A weather station wants to know how many days were warmer than average.

### Steps:

    Create temps = [30, 32, 28, 35, 31, 29, 36].
    
    First loop: add all the temperatures and calculate the average.
    
    Second loop: count the days above the average.
    
    Print the average (2 decimals) and the count.

### Example:

    Average = 31.57
    Days above average = 3

```python
temps = [30, 32, 28, 35, 31, 29, 36]

# First loop: Calculate the total temperature
total = 0
for temp in temps:
    total += temp

average = total / len(temps)

# Second loop: Count days above average
count = 0
for temp in temps:
    if temp > average:
        count += 1

# Print the results
print("Average temperature:", round(average, 2))
print("Days above average:", count)
```
