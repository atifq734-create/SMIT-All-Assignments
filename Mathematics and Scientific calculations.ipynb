# 7) Compound Interest:
Write a Python program to calculate compound interest. Use the formula:

    A=p*(1+r/n)**(n-t)

Where:

A= total amount

P = principal amount

r = annual interest rate (decimal)

n = number of times interest is compounded per year

t = time in years

Take P, r, n, and t as inputs from the user.

```python
p = float(input("Enter principal amount (P): "))
r = float(input("Enter annual interest rate (decimal): "))
n = int(input("Enter number of times interest is compounded per year (n): "))
t = float(input("Enter time in years (t): "))

A = p * (1 + r / n) ** (n - t)

print("Total Amount (A) =", A)
```

# 11) Quadratic Equation Roots:

Write a Python program to find the roots of a quadratic equation. Use the formula:

X=

Take a, b, and c as inputs from the user. 


```python
import math

a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

discriminant = b ** 2 - 4 * a * c

if discriminant >= 0:
    x1 = (-b + math.sqrt(discriminant)) / (2 * a)
    x2 = (-b - math.sqrt(discriminant)) / (2 * a)

    print("Root 1 =", x1)
    print("Root 2 =", x2)
else:
    print("The equation has no real roots.")
```

# 19) Work Done by a Force:

Write a Python program to calculate the work done by a force. 
Use the formula: 
W=fd cos


    where:

    W = work done

f= force

d = distance

theta = angle between force and direction of movement (in degrees)

Take f, d, and θ as inputs from the user.


```python
import math

f = float(input("Enter force (N): "))
d = float(input("Enter distance (m): "))
theta = float(input("Enter angle (degrees): "))

work = f * d * math.cos(math.radians(theta))

print("Work Done =", work, "J")
```

# 20) Heat Transfer:

Write a Python program to calculate the amount of heat transferred. 
Use the formula:

𝑄 = 𝑚𝑐 (∆T)

where:

Q= heat transfer

m = mass

c = specific heat capacity

ΔT

Take m, c, T as inputs from the user.

```python
m = float(input("Enter mass (kg): "))
c = float(input("Enter specific heat capacity (J/kg°C): "))
T1 = float(input("Enter initial temperature (°C): "))
T2 = float(input("Enter final temperature (°C): "))

delta_T = T2 - T1

Q = m * c * delta_T

print("Heat transferred =", Q, "J")
```
